```python
import pandas as pd
from pathlib import Path

# --- CONFIG: adjust if your filename/path is different ---
csv_path = Path("logdata/mupromac_log_FIFO_run0_EXP_dedicated_C1.csv")

expected_cols = {
    "Case ID",
    "Activity",
    "Resource",
    "Complete Timestamp",
    "activity_duration",
    "timesincelastevent",
    "timesincecasestart",
    "event_nr",
    "open_cases",
    "remtime",
}

print(f"Checking: {csv_path.resolve()}")

# 1) Try reading with ; separator (what DatasetManager uses)
try:
    df = pd.read_csv(csv_path, sep=";")
except Exception as e:
    print("❌ Could not read CSV with sep=';':", e)
    raise SystemExit

print(f"✅ Loaded CSV with sep=';': {df.shape[0]} rows, {df.shape[1]} columns")

actual_cols = set(df.columns)
print("\nColumns in file:")
for c in df.columns:
    print("  -", c)

missing = expected_cols - actual_cols
extra   = actual_cols - expected_cols

print("\n--- Column check ---")
if not missing:
    print("✅ All expected columns are present.")
else:
    print("❌ Missing columns:", missing)

if extra:
    print("ℹ️ Extra columns (OK, just unused by the benchmark):", extra)

# 2) Quick type check for numeric columns
numeric_cols = [
    "activity_duration",
    "timesincelastevent",
    "timesincecasestart",
    "event_nr",
    "open_cases",
    "remtime",
]

print("\n--- Numeric type check ---")
for col in numeric_cols:
    if col not in df.columns:
        print(f"  ⚠️ {col} not found (already reported above)")
        continue
    # Try to coerce to numeric
    coerced = pd.to_numeric(df[col], errors="coerce")
    n_na = coerced.isna().sum()
    if n_na == 0:
        print(f"  ✅ {col} looks numeric.")
    else:
        print(f"  ⚠️ {col} has {n_na} non-numeric value(s).")

print("\nDone.")
```
