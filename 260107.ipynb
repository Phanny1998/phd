```python
import pandas as pd
df_raw = pd.read_csv("out/251110/results/FIFO_EXP_l0.28_dedicated_C1_V6_A12_QC100_mild_all.csv")

# Filter to running status
df_running = df_raw[df_raw["status"] == "running"]

# Get all inspection activities
inspection_acts = [act for act in df_running['activity'].unique() if 'INSPECTION' in act]
print("Inspection activities in running events:")
print(sorted(inspection_acts))
```
